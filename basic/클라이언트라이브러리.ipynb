```python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)

openai_api_key = os.getenv('OPENAI_API_KEY')
anthropic_api_key = os.getenv('ANTHROPIC_API_KEY')
google_api_key = os.getenv('GOOGLE_API_KEY')
deepseek_api_key = os.getenv('DEEPSEEK_API_KEY')
groq_api_key = os.getenv('GROQ_API_KEY')
grok_api_key = os.getenv('GROK_API_KEY')
openrouter_api_key = os.getenv('OPENROUTER_API_KEY')
```

```python
# OpenAI 클라이언트 라이브러리에 연결
# HTTP 엔드포인트 호출을 감싸는 얇은 래퍼입니다

openai = OpenAI()

# Gemini, DeepSeek, Groq의 경우 OpenAI 파이썬 클라이언트를 사용할 수 있습니다
# Google과 DeepSeek이 OpenAI와 호환되는 엔드포인트를 제공하기 때문입니다
# 그리고 OpenAI 라이브러리는 base_url을 변경할 수 있게 해줍니다

anthropic_url = "https://api.anthropic.com/v1/"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
deepseek_url = "https://api.deepseek.com"
groq_url = "https://api.groq.com/openai/v1"
grok_url = "https://api.x.ai/v1"
openrouter_url = "https://openrouter.ai/api/v1"
ollama_url = "http://localhost:11434/v1"

anthropic = OpenAI(api_key=anthropic_api_key, base_url=anthropic_url)
gemini = OpenAI(api_key=google_api_key, base_url=gemini_url)
deepseek = OpenAI(api_key=deepseek_api_key, base_url=deepseek_url)
groq = OpenAI(api_key=groq_api_key, base_url=groq_url)
grok = OpenAI(api_key=grok_api_key, base_url=grok_url)
openrouter = OpenAI(base_url=openrouter_url, api_key=openrouter_api_key)
ollama = OpenAI(api_key="ollama", base_url=ollama_url)
```

## 학습 시간 대 추론 시간 스케일링

```python
easy_puzzle = [
    {"role": "user", "content":
        "동전을 2개 던졌습니다. 그중 하나는 앞면입니다. 나머지 하나가 뒷면일 확률은 얼마인가요? 확률만 답하세요."},
]
```

```python
response = openai.chat.completions.create(model="gpt-5-nano", messages=easy_puzzle, reasoning_effort="minimal")
display(Markdown(response.choices[0].message.content))
```

```python
response = openai.chat.completions.create(model="gpt-5-nano", messages=easy_puzzle, reasoning_effort="low")
display(Markdown(response.choices[0].message.content))
```

```python
response = openai.chat.completions.create(model="gpt-5-mini", messages=easy_puzzle, reasoning_effort="minimal")
display(Markdown(response.choices[0].message.content))
```

## Gemini와 Anthropic 클라이언트 라이브러리

지금까지는 OpenAI 파이썬 클라이언트 라이브러리를 경유했지만, 다른 제공업체들도 각자의 라이브러리를 가지고 있습니다

```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)
```

```python
from google import genai

google_api_key = os.getenv('GOOGLE_API_KEY')

client = genai.Client()

response = client.models.generate_content(
    model="gemini-3.1-flash-lite", contents="파란색을 한 번도 본 적 없는 사람에게 한 문장으로 설명해줘"
)

print(response.text)
```

```python
from anthropic import Anthropic

anthropic_api_key = os.getenv('ANTHROPIC_API_KEY')

client = Anthropic()

response = client.messages.create(
    model="claude-sonnet-4-5-20250929",
    messages=[{"role": "user", "content": "파란색을 한 번도 본 적 없는 사람에게 한 문장으로 설명해줘"}],
    max_tokens=100
)
print(response.content[0].text)
```
