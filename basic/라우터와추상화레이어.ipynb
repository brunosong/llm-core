## 라우터와 추상화 레이어

```python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display


openrouter_api_key = os.getenv('OPENROUTER_API_KEY')
openrouter_url = "https://openrouter.ai/api/v1"

openrouter = OpenAI(base_url=openrouter_url, api_key=openrouter_api_key)
```

```python

tell_a_joke = [
    {"role": "user", "content": "LLM 엔지니어링 전문가가 되기 위한 여정을 걷고 있는 학생을 위한 농담을 하나 해줘"},
]


response = openrouter.chat.completions.create(model="z-ai/glm-4.5", messages=tell_a_joke)
display(Markdown(response.choices[0].message.content))
```

# LangChain

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-5-mini")
response = llm.invoke(tell_a_joke)

display(Markdown(response.content))
```

# LiteLLM

```python
from litellm import completion

# 모델 명을 넣으면 알아서 회사별 API KEY 를 찾아서 API 콜
# 지금 같이 OPEN AI꺼를 사용하면 OPEN_API_KEY env가 있어야 함
response = completion(model="openai/gpt-4.1", messages=tell_a_joke)
reply = response.choices[0].message.content
display(Markdown(reply))
```

```python
print(f"입력 토큰 수: {response.usage.prompt_tokens}")
print(f"출력 토큰 수: {response.usage.completion_tokens}")
print(f"총 토큰 수: {response.usage.total_tokens}")
print(f"총 비용: {response._hidden_params["response_cost"]*100:.4f} 센트")
```

# 프롬프트 캐싱

```python
with open("hamlet.txt", "r", encoding="utf-8") as f:
    hamlet = f.read()

loc = hamlet.find("Speak, man")
print(hamlet[loc:loc+100])
```

```python
question = [{"role": "user", "content": "햄릿에서 레어티스가 '제 아버지는 어디 계십니까?'라고 물었을 때 대답은 무엇인가요?"}]
```

```python
response = completion(model="gemini/gemini-3.1-flash-lite", messages=question)
display(Markdown(response.choices[0].message.content))
```

```python
print(f"입력 토큰 수: {response.usage.prompt_tokens}")
print(f"출력 토큰 수: {response.usage.completion_tokens}")
print(f"총 토큰 수: {response.usage.total_tokens}")
print(f"총 비용: {response._hidden_params["response_cost"]*100:.4f} 센트")
```

```python
question[0]["content"] += "\n\n참고로, 다음은 햄릿 전체 텍스트입니다:\n\n"+hamlet
```

```python
response = completion(model="gemini/gemini-3.1-flash-lite", messages=question)
display(Markdown(response.choices[0].message.content))
```

```python
print(f"입력 토큰 수: {response.usage.prompt_tokens}")
print(f"출력 토큰 수: {response.usage.completion_tokens}")
print(f"캐시된 토큰 수: {response.usage.prompt_tokens_details.cached_tokens}")
print(f"총 비용: {response._hidden_params["response_cost"]*100:.4f} 센트")
```
