```python
import tiktoken

encoding = tiktoken.encoding_for_model("gpt-4.1-mini")

tokens = encoding.encode("Hi my name is brunosong and I like pie")
```

```python
tokens
```

```python
for token_id in tokens:
    token_text = encoding.decode([token_id])
    print(f"{token_id} = {token_text}")
```

```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)
api_key = os.getenv('GOOGLE_API_KEY')
```

```python
from openai import OpenAI

GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"

gemini = OpenAI(
    base_url = GEMINI_BASE_URL,
    api_key = api_key
)
```

```python

messages = [
    { "role" : "system" , "content" : "너는 도움이 되는 보조자 입니다. "},
    {"role": "user", "content": "안녕 나에 이름은 브루노송이야."}
]
```

```python
response = gemini.chat.completions.create(model="gemini-3.1-flash-lite", messages=messages)
response.choices[0].message.content
```
