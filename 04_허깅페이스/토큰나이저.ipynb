```python
#from google.colab import userdata
from huggingface_hub import login
from transformers import AutoTokenizer
import os
from dotenv import load_dotenv
```

```python

hf_token = os.getenv('HF_TOKEN')
if hf_token and hf_token.startswith("hf_"):
  print("HF key looks good so far")
else:
  print("HF key is not set - please click the key in the left sidebar")
login(hf_token, add_to_git_credential=True)
```

```python
tokenizer = AutoTokenizer.from_pretrained('meta-llama/Meta-Llama-3.1-8B', trust_remote_code=True)
```

```python
text = "LLM 엔지니어들에게 Tokenizers가 실제로 작동하는 모습을 보여줄 수 있게 되어 기대가 큽니다."
tokens = tokenizer.encode(text)  # 토큰 ID로 변경한다.
tokens
```

```python
character_count = len(text)
word_count = len(text.split(' '))
token_count = len(tokens)
print(f"There are {character_count} characters, {word_count} words and {token_count} tokens")
```

```python
tokenizer.decode(tokens)
```

```python
tokenizer.batch_decode(tokens)
```

# 기본 모델
이제 한 걸음 더 나아가서, 우리가 살펴본 이 라마 모델은 베이스모델 이라고 부르는 것의 한 예시입니다.
베이스 모델, 채팅 모델, 추론(reasoning) 모델 등 여러 종류가 있습니다. 지금 위에서 사용한 모델은 베이스 모델이기 때문에, 시스템 프롬프트나 사용자 프롬프트 같은 형식을 기대하지 않습니다. 애초에 학습 데이터 자체에 그런 형식(대화 형식)이 없었기 때문이죠.

그래서 베이스 모델은 상당히 "빈 캔버스"에 가깝습니다. 물론 방대한 학습 데이터를 이미 흡수했기 때문에 오나전히 텅 빈 상태는 아니지만, 다른 모델들에 비하면 훨씬 더 "비어있는" 상태이고, 원하는 방향으로 자유롭게 활용할 수 있도록 준비된 상태라고 보시면 됩ㄴ디ㅏ.

이런 베이스 모델에는 "채팅 변형" 또는 인스트럭트 변형 이라고 불리는 버전이 존재합니다. 이 변형들은 우리가 지난 몇 주간 계속 사용해온 형태, 즉 시스템 프롬프트, 사용자 프롬프트, 어시스턴트 응답이 앞뒤로 오가는 대화형식 을 처리하도록 특별히 추가 학습된 모델 입니다.

참고로 허깅페이스는 보통 채팅 이라는 단어보다 Instruct라는 단어를 더 많이 사용하는 편이지만, 두 용어 모두 흔히 볼 수 있습니다.

이 Instracut 변형 모델들은 앞서 말한 대화 형식의 프롬프트를 받도록 특별히 파인튜닝 된 것들입니다. 그리고 허깅페이스 개발자들은 'apply_chat_template'이라는 유용한 유틸리티 함수를 만들었는데, 이 함수는 OpenAI API 방식처럼 우리에게 익숙한 딕셔너리 리스트 형태의 입력을 받아서 해당 모델에 맞는 특수 토큰과 텍스트 형식으로 자동 변환해 줍니다.



# 모델의 Instruct 변형

많은 모델에는 채팅에서 사용하도록 학습된 변형이 존재합니다. 이러한 변형은 보통 이름 끝에 "Instruct" 라는 단어가 붙습니다.
이 모델들을 system,user,assistant 프롬프트를 포함하는 특정 형식의 프롬프트를 입력받도록 학습되었습니다.

우리에게 익숙한 messages 리스트 형식을 이 모델에 맞는 올바른 입력 프롬프트로 변환해주는 'apply_chat_template' 이라는 유틸리티 메서드가 있습니다.
